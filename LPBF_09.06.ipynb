```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.MultiDiGraph()
fig = plt.figure(figsize=(2,2),dpi=300)

G.add_node("LPBF", shape='square')
G.add_nodes_from(["20-100", "20-75", "20-125", "20-150", "15-100", "15-150", "15-75", "25-100", "30-100", "30-75", "30-150"])

# Add edges to the graph
#G.add_edge("LPBF", "20-100")
#G.add_edge("20-100", "LPBF")
G.add_edges_from([("LPBF", node) for node in G.nodes() if node != "LPBF"])
G.add_edges_from([(node, "LPBF") for node in G.nodes() if node != "LPBF"])

# Calculate the positions of nodes using the Fruchterman-Reingold layout algorithm
pos = nx.spring_layout(G, center=(0, 0), seed=42)

# Visualize the graph
# nx.draw_networkx(G, pos, with_labels=True, node_color='darkred', node_size=1000, font_weight='bold', connectionstyle='arc3,rad=0.2')
nx.draw_networkx_nodes(G, pos, node_color='royalblue', node_size=200, nodelist=['LPBF'], node_shape='s')
nx.draw_networkx(G, pos, with_labels=False, node_color='darkred', node_size=200, nodelist=[node for node in G.nodes() if node != 'LPBF'], node_shape='o', connectionstyle='arc3,rad=0.15')

plt.savefig('cluster')
plt.axis("off")
plt.show()
```
