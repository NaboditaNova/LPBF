```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.MultiDiGraph()

G.add_node("LPBF")
G.add_nodes_from(["20-100", "20-75", "20-125", "20-150", "15-100", "15-150", "15-75", "25-100", "30-100", "30-75", "30-150"])

# Add edges to the graph
G.add_edge("LPBF", "20-100")
G.add_edge("20-100","LPBF")
G.add_edges_from([("LPBF",node) for node in G.nodes() if node !="20-100"])
G.add_edges_from([(node, "LPBF")for node in G.nodes()if node !="20-100"])
pos = nx.circular_layout(G)

# Adjust the positions of the edges
pos_adjusted = pos.copy()
for edge in G.edges():
    if edge[0] == "LPBF":
        pos_adjusted[edge[0]] += [0.05, 0.05]
    elif edge[1] == "LPBF":
        pos_adjusted[edge[1]] += [-0.05, 0.05]

# Visualize the graph
nx.draw_networkx(G, pos_adjusted, with_labels=True, node_color='lightblue', node_size=1000, font_weight='bold', connectionstyle='arc3,rad=0.1')

plt.show()
```

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.MultiDiGraph()

G.add_node("LPBF")
G.add_nodes_from(["20-100", "20-75", "20-125", "20-150", "15-100", "15-150", "15-75", "25-100", "30-100", "30-75", "30-150"])

# Add edges to the graph
G.add_edge("LPBF", "20-100")
G.add_edges_from([("LPBF", node) for node in G.nodes() if node != "20-100"])
G.add_edges_from([(node, "LPBF") for node in G.nodes() if node != "20-100"])

# Set the positions of nodes
pos = nx.circular_layout(G)

# Set the center node
center_node = "LPBF"

# Adjust the positions of the nodes
pos_adjusted = pos.copy()
for node in pos_adjusted:
    if node != center_node:
        delta_x = pos_adjusted[node][0] - pos_adjusted[center_node][0]
        delta_y = pos_adjusted[node][1] - pos_adjusted[center_node][1]
        pos_adjusted[node] = (pos_adjusted[center_node][0] + delta_x * 1.2, pos_adjusted[center_node][1] + delta_y * 1.2)

# Visualize the graph
nx.draw_networkx(G, pos_adjusted, with_labels=True, node_color='lightblue', node_size=1000, font_weight='bold')

plt.show()

```

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.MultiDiGraph()

G.add_node("LPBF")
G.add_nodes_from(["20-100", "20-75", "20-125", "20-150", "15-100", "15-150", "15-75", "25-100", "30-100", "30-75", "30-150"])

# Add edges to the graph
G.add_edge("LPBF", "20-100")
G.add_edge("20-100", "LPBF")
G.add_edges_from([("LPBF", node) for node in G.nodes() if node != "20-100"])
G.add_edges_from([(node, "LPBF") for node in G.nodes() if node != "20-100"])

# Set the positions of nodes
pos = nx.circular_layout(G)

# Set the position of the LPBF node to the center
pos["LPBF"] = [0, 0]

# Adjust the positions of the other nodes
for node, position in pos.items():
    if node != "LPBF":
        angle = nx.circular_layout(G)[node]
        radius = 1.5
        pos[node] = [radius * position[0] + angle[0], radius * position[1] + angle[1]]

# Visualize the graph
nx.draw_networkx(G, pos, with_labels=True, node_color='lightblue', node_size=1000, font_weight='bold', connectionstyle='arc3,rad=0.2')

plt.axis("off")
plt.show()
```
